```python
import requests
import numpy as np
import pandas as pd
import matplotlib
import seaborn
from pyjstat import pyjstat

base_url = "https://data.ssb.no/api/pxwebapi/v2/tables/07230/data?lang=no&outputFormat=json-stat2&valuecodes[ContentsCode]=*&valuecodes[Tid]=*&valuecodes[Region]=*&valuecodes[Boligtype]=*&heading=ContentsCode,Tid,Boligtype&stub=Region"

datasett = pyjstat.Dataset.read(base_url)
df = datasett.write("dataframe")
# 48 X 5 - rad: region, kolonner: informajon
print(df.head(10))
print("Nan values",df["value"].isna().sum())
```

1) Fyller på Nan deler for bolig indeksene 

```python
arrLen = np.mean(df["value"])
# erstatter Nan med gjennomsnitt verdien
df["value"] = df["value"].fillna(arrLen)
print("Nan values",df["value"].isna().sum())
```
